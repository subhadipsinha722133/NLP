```python
import pandas as pd
a=pd.read_csv(r"D:\CODING\PYTHON\NLP\5.sms-spam-classifier-main\spam.csv",encoding="cp1252")
a
```

```python
a=a.iloc[:,:-3]
a
```

```python
import re
def remove_html(text):
    p=re.compile("<.?>")
    return p.sub(r"",text)

a["v2"]=a["v2"].apply(remove_html)
```

```python
a["v2"]=a["v2"].str.lower()
a
```

```python
import string
pu=string.punctuation

def remove_punctuation(text):
    return text.translate(str.maketrans("","",pu))

a["v2"]=a["v2"].apply(remove_punctuation)
a
```

```python
from nltk.corpus import stopwords

def remove_stopword(text):
    new_text=[]
    for i in text.split():
        if i in stopwords.words("english"):
            new_text.append("")
        else:
            new_text.append(i)
    x=new_text[ : ]
    new_text.clear()
    return " ".join(x)

a["v2"]=a["v2"].apply(remove_stopword)
```

```python
from sklearn.preprocessing import LabelEncoder

la=LabelEncoder()
y=la.fit_transform(a["v1"])
y
```

```python
from sklearn.feature_extraction.text import  TfidfVectorizer

tfidf=TfidfVectorizer()
x=tfidf.fit_transform(a["v2"]).toarray()
# x_test=tfidf.fit_transform(x_test).toarray()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
gnb=RandomForestClassifier()
gnb.fit(x_train,y_train)
gnb.score(x_test,y_test)*100
```
