```python
##tensorflow >2.0
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good']
```

```python
sent
```

```python
### Vocabulary size
voc_size=500
```

#### One Hot Representation

```python
onehot_repr=[one_hot(words,voc_size)for words in sent]
print(onehot_repr)
```

### Word Embedding Represntation

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
## pre padding
sent_length = 8
embedded_docs = pad_sequences(onehot_repr, padding='pre', maxlen=sent_length)
print("Padded sequences:","\n", embedded_docs)
```

```python
## 10 feature dimesnions
dim=10
```

```python
# model=Sequential()
# model.add(Embedding(voc_size,10,input_length=sent_length))
# model.compile('adam','mse')


model = Sequential()
model.add(Embedding(voc_size, 10, input_length=sent_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
##'the glass of milk',
embedded_docs[0]
```

```python
# CORRECT WAY: Add batch dimension for single prediction
single_sample = embedded_docs[0].reshape(1, -1)  # Shape: (1, 8)
print("Single sample shape:", single_sample.shape)
```

```python
model.predict(embedded_docs[0])
```

```python
# Now this will work
prediction = model.predict(single_sample)
print("Prediction shape:", prediction.shape)
print("Prediction for first sample:")
print(prediction)
```

```python
# Predict on all samples (already has correct shape: (7, 8))
all_predictions = model.predict(embedded_docs)
print("\nAll predictions shape:", all_predictions.shape)
print("All predictions:")
print(all_predictions)
```
